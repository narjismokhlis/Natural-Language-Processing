This is a notebook for a 'smart' tokenizer I want to build. It's supposed to be based on machine learning, specifically logistic regression (supervised learning). The most tricky part of toneizing a text is the preprocessing part, which include sentence segmentation. This is the part that requires binary classification using logistic regression. The model should be able to predict whether a space in the text indicates the end of a sentence, or not.

The training data will consist of:
- A segment of text where the space indicates a sentence boundary (labeled 1).
- A segment of text where the space does not indicate a sentence boundary (labeled 0).
